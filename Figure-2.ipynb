```python
def data_to_istance_lists(data):
    out_dict = {}
    for key in data.keys():
        out_dict[(key[0],key[1],key[2])] = []
    for key in data.keys():
        out_dict[(key[0],key[1],key[2])].append(data[key])

    return out_dict


def take_N(data,N):
    taken = {}
    for key in data.keys():
        if key[0] == N:
            taken[(key[1], key[2])] = data[key]
    return taken


def take_average(data):
    averaged = {}
    for key in data.keys():
        averaged[key] = np.mean(data[key])
    return averaged


def take_first_element(data):
    out_dict = {}
    for key in data.keys():
        out_dict[key] = data[key][0]
    return out_dict


def take_second_element(data):
    out_dict = {}
    for key in data.keys():
        out_dict[key] = data[key][1]
    return out_dict
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ── Your data ──────────────────────────────────────────────

# Add more dictionaries here: e.g. diag = {...}

dicts = {
    "Lower bound": ER_ER_lower,
    "Umeyama": ER_ER_UMEYAMA,
    "Degree": ER_ER_degree,
    "GRAMPA": ER_ER_aligned,
    "GRASP": ER_ER_GRASP,
    "Random": ER_ER_random,
    "Upper bound": ER_ER_upper,
}

# ── Layout ─────────────────────────────────────────────────
keys       = [(2,2), (2,4), (2,6),(2,8),(4,4),(4,6),(4,8),(6,6),(6,8),(8,8)]
col_labels =    ["(2,2)", "(2,4)", "(2,6)","(2,8)","(4,4)","(4,6)","(4,8)","(6,6)","(6,8)","(8,8)"] 
row_labels = list(dicts.keys())

z = np.array([[dicts[row][k] for k in keys] for row in row_labels])

# ── Plot ───────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 3.5))
im = ax.imshow(z, cmap="viridis", aspect="auto", vmin=0, vmax=3)

# Annotate each cell
for i in range(len(row_labels)):
    for j in range(len(col_labels)):
        val = z[i, j]
        color = "white" if val > 0 else "black"
        ax.text(j, i, f"{val:.2f}", ha="center", va="center",
                fontsize=8, fontweight="bold", color=color)

ax.set_xticks(range(len(col_labels)))
ax.set_xticklabels(col_labels, fontsize=8)
ax.set_yticks(range(len(row_labels)))
ax.set_yticklabels(row_labels, fontsize=8)
ax.set_xlabel("Approximate average degrees $(\\langle k_1 \\rangle, \\langle k_2 \\rangle)$", fontsize=10)
ax.set_ylabel("Node-correspondence", fontsize=10)
ax.set_title("Average normalized GGD of ER vs ER", fontsize=12, fontweight="bold", pad=12)

cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.ax.tick_params(labelsize=11)

plt.tight_layout()
plt.savefig("figures/Fig2_ER_ER.png", dpi=150, bbox_inches="tight")
plt.show()
```
