```python
#Dependence on N plot
import numpy as np
import matplotlib.pyplot as plt
import pickle
```

```python
def process_data(data):
    out_dict = {}
    for key in data.keys():
        out_dict[(key[0],key[1],key[2])] = []
    for key in data.keys():
        out_dict[(key[0],key[1],key[2])].append(data[key])

    for key in out_dict.keys():
        out_dict[key] = (np.mean(np.array(out_dict[key])), np.std(np.array(out_dict[key])))
    
    return out_dict


def take_first_element(data):
    out_dict = {}
    for key in data.keys():
        out_dict[key] = data[key][0]
    return out_dict


def take_second_element(data):
    out_dict = {}
    for key in data.keys():
        out_dict[key] = data[key][1]
    return out_dict
```

```python
#load data
with open('data/ggd_calculation_results/ER_GRAMPA.pkl', 'rb') as f:
    ER_aligned = pickle.load(f)

with open('data/ggd_calculation_results/BA_GRAMPA.pkl', 'rb') as f:
    BA_aligned = pickle.load(f)

with open('data/ggd_calculation_results/ER_BA_GRAMPA.pkl', 'rb') as f:
    ER_BA_aligned = pickle.load(f)
    
with open('data/ggd_calculation_results/ER_random.pkl', 'rb') as f:
    ER_random = pickle.load(f)

with open('data/ggd_calculation_results/BA_random.pkl', 'rb') as f:
    BA_random = pickle.load(f)

with open('data/ggd_calculation_results/ER_BA_random.pkl', 'rb') as f:
    ER_BA_random = pickle.load(f)

with open('data/ggd_calculation_results/ER_degree.pkl', 'rb') as f:
    ER_degree = pickle.load(f)

with open('data/ggd_calculation_results/BA_degree.pkl', 'rb') as f:
    BA_degree = pickle.load(f)

with open('data/ggd_calculation_results/ER_BA_degree.pkl', 'rb') as f:
    ER_BA_degree = pickle.load(f)

with open('data/ggd_calculation_results/ER_bounds.pkl', 'rb') as f:
    ER_bounds = pickle.load(f)

with open('data/ggd_calculation_results/BA_bounds.pkl', 'rb') as f:
    BA_bounds = pickle.load(f)

with open('data/ggd_calculation_results/ER_BA_bounds.pkl', 'rb') as f:
    ER_BA_bounds = pickle.load(f)

with open('data/ggd_calculation_results/ER_GRASP.pkl', 'rb') as f:
    ER_GRASP = pickle.load(f)

with open('data/ggd_calculation_results/BA_GRASP.pkl', 'rb') as f:
    BA_GRASP = pickle.load(f)

with open('data/ggd_calculation_results/ER_BA_GRASP.pkl', 'rb') as f:
    ER_BA_GRASP = pickle.load(f)

with open('data/ggd_calculation_results/ER_UMEYAMA.pkl', 'rb') as f:
    ER_UMEYAMA = pickle.load(f)

with open('data/ggd_calculation_results/BA_UMEYAMA.pkl', 'rb') as f:
    BA_UMEYAMA = pickle.load(f)

with open('data/ggd_calculation_results/ER_BA_UMEYAMA.pkl', 'rb') as f:
    ER_BA_UMEYAMA = pickle.load(f)

with open('data/ggd_calculation_results/BA_natural.pkl', 'rb') as f:
    BA_natural = pickle.load(f)
```

```python
ER_ER_grampa_ms = process_data(ER_aligned)
ER_ER_grasp_ms = process_data(ER_GRASP)
ER_ER_umeyama_ms = process_data(ER_UMEYAMA)
ER_ER_degree_ms = process_data(ER_degree)
ER_ER_random_ms = process_data(ER_random)
ER_ER_lower_ms = process_data(take_first_element(ER_bounds))
ER_ER_upper_ms = process_data(take_second_element(ER_bounds))

BA_BA_grampa_ms = process_data(BA_aligned)
BA_BA_grasp_ms = process_data(BA_GRASP)
BA_BA_umeyama_ms = process_data(BA_UMEYAMA)
BA_BA_degree_ms = process_data(BA_degree)
BA_BA_random_ms = process_data(BA_random)
BA_BA_lower_ms = process_data(take_first_element(BA_bounds))
BA_BA_upper_ms = process_data(take_second_element(BA_bounds))
BA_BA_natueral_ms = process_data(BA_natural)

ER_BA_grampa_ms = process_data(ER_BA_aligned)
ER_BA_grasp_ms = process_data(ER_BA_GRASP)
ER_BA_umeyama_ms = process_data(ER_BA_UMEYAMA)
ER_BA_degree_ms = process_data(ER_BA_degree)
ER_BA_random_ms = process_data(ER_BA_random)
ER_BA_lower_ms = process_data(take_first_element(ER_BA_bounds))
ER_BA_upper_ms = process_data(take_second_element(ER_BA_bounds))


```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np





# ── Your data ──────────────────────────────────────────────────────────────────
# Each batch is a list of dicts: [{(N, k1, k2): (mean, std), ...}, ...]
# Row 0: 8 sub-batches; Rows 1 & 2: 7 sub-batches

N_vals = [512, 1024, 2048]

# (k1, k2) pairs per column; same across rows if desired
row0_k_pairs = [(2, 2), (2, 8), (8, 8)]
row1_k_pairs = [(2, 2), (2, 8), (8, 8)]
row2_k_pairs = [(2, 2), (2, 8), (8, 8)]


row0_batches = [BA_BA_lower_ms, 
                BA_BA_umeyama_ms,
                BA_BA_degree_ms,
                BA_BA_grampa_ms,
                BA_BA_grasp_ms,
                BA_BA_random_ms,
                BA_BA_natueral_ms,
                BA_BA_upper_ms]  # length 8

row1_batches = [ER_BA_lower_ms, 
                ER_BA_umeyama_ms,
                ER_BA_degree_ms,
                ER_BA_grampa_ms,
                ER_BA_grasp_ms,
                ER_BA_random_ms,
                ER_BA_upper_ms]  # length 7

row2_batches = [ER_ER_lower_ms, 
                ER_ER_umeyama_ms,
                ER_ER_degree_ms,
                ER_ER_grampa_ms,
                ER_ER_grasp_ms,
                ER_ER_random_ms,
                ER_ER_upper_ms]  # length 7

# (batches, k_pairs, number_of_lines, row_label)
#"Barabási-Albert vs Barabási-Albert"
#"Erdős-Rényi vs Barabási-Albert"
#"Erdős-Rényi vs Erdős-Rényi"
rows = [
    (row0_batches, row0_k_pairs, 8, "BA vs BA"),
    (row1_batches, row1_k_pairs, 7, "ER vs BA"),
    (row2_batches, row2_k_pairs, 7, "ER vs ER"),
]

# Column titles — same for all rows
col_titles = [
    '$\\langle k_1 \\rangle \\approx 2,\\ \\langle k_2 \\rangle \\approx 2$',
    '$\\langle k_1 \\rangle \\approx 2,\\ \\langle k_2 \\rangle \\approx 8$',
    '$\\langle k_1 \\rangle \\approx 8,\\ \\langle k_2 \\rangle \\approx 8$',
]

# Legend labels per row (adjust as needed)
labels_row0   = ["Lower bound", "Umeyama", "Degree", "GRAMPA",
                 "GRASP", "Random", "Natural", "Upper bound"]
labels_rows12 = ["Lower bound", "Umeyama", "Degree", "GRAMPA",
                 "GRASP", "Random", "Upper bound"]

# ── Colors: black for first/last, distinct colors in between ───────────────────
COLORS = ["#E63946", "#2196F3", "#FF9800", "#4CAF50", "#9C27B0", "#00BCD4"]

row_colors = [
    ["black"] + COLORS[:6] + ["black"],  # row 0: 8 lines
    ["black"] + COLORS[:5] + ["black"],  # row 1: 7 lines
    ["black"] + COLORS[:5] + ["black"],  # row 2: 7 lines
]


def get_ls(sub_idx, n_sub):
    if sub_idx == 0:
        return "-"
    elif sub_idx == n_sub - 1:
        return "-"
    return "-"

# ── Figure ────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(3, 3, figsize=(11, 8), sharey="row", sharex=True)
fig.subplots_adjust(hspace=0.08, wspace=0.06, left=0.10, right=0.72, top=0.93, bottom=0.18)

x_ticks = [512, 1024, 2048]

for row_idx, (batches, k_pairs, n_sub, row_label) in enumerate(rows):
    labels = labels_row0 if row_idx == 0 else labels_rows12
    colors = row_colors[row_idx]
    legend_handles = []

    for col_idx, (k1, k2) in enumerate(k_pairs):
        ax = axes[row_idx][col_idx]

        for sub_idx, batch in enumerate(batches):
            means = np.array([batch[(N, k1, k2)][0] for N in N_vals])
            stds  = np.array([batch[(N, k1, k2)][1] for N in N_vals])
            color = colors[sub_idx]
            ls    = get_ls(sub_idx, n_sub)

            line, _, _ = ax.errorbar(
                N_vals, means,
                yerr=stds,
                fmt='-',
                marker=None,
                color=color,
                linestyle=ls,
                linewidth=1.5,
                capsize=3,
                capthick=1,
                elinewidth=1.2,
            )

            if col_idx == 0:
                legend_handles.append(line)

        ax.set_xscale("log", base=2)
        ax.set_xticks(x_ticks)
        ax.set_xlim(N_vals[0], N_vals[-1])
        ax.xaxis.set_minor_locator(ticker.NullLocator())
        ax.yaxis.set_minor_locator(ticker.NullLocator())
        ax.tick_params(axis="y", which="major", labelsize=11)

        # x tick labels in ALL bottom-row columns, vertical to avoid overlap
        if row_idx == 2:
            ax.xaxis.set_major_formatter(ticker.FixedFormatter(["512", "1024", "2048"]))
            ax.tick_params(axis="x", labelbottom=True, labelsize=8)
            for lbl in ax.get_xticklabels():
                lbl.set_rotation(90)
                lbl.set_ha("center")
                lbl.set_va("top")
        else:
            ax.tick_params(axis="x", labelbottom=False)

        if row_idx == 0:
            ax.set_title(col_titles[col_idx], fontsize=12, pad=4)

        if col_idx == 2:
            ax.yaxis.set_label_position("right")
            ax.set_ylabel(row_label, fontsize=13, rotation=270, labelpad=8, va="bottom")

    axes[row_idx][2].legend(
        handles=legend_handles,
        labels=labels,
        title="Node-correspondence",
        fontsize=9,
        title_fontsize=10,
        loc="upper left",
        bbox_to_anchor=(1.35, 1.0),
        borderaxespad=0,
        ncol=1,
        labelspacing=0.2,
        handlelength=1.5,
        handletextpad=0.4,
    )

# ── Shrink column widths tightly (factor 0.7) ────────────────────────────────
n_rows, n_cols = axes.shape
shrink = 0.7
positions = [[ax.get_position() for ax in row] for row in axes]
new_width = positions[0][0].width * shrink
left_start = positions[0][0].x0
gap = 0.01

for col_idx in range(n_cols):
    new_x0 = left_start + col_idx * (new_width + gap)
    for row_idx in range(n_rows):
        pos = positions[row_idx][col_idx]
        axes[row_idx][col_idx].set_position([new_x0, pos.y0, new_width, pos.height])

# ── Shared axis labels ────────────────────────────────────────────────────────
fig.text(0.33, 0.08, "Number of nodes", ha="center", fontsize=14)
fig.text(0.03, 0.55, "Average normalized GGD", va="center", rotation="vertical", fontsize=14)

plt.suptitle("", fontsize=16)
plt.savefig("figures/Fig1.png", dpi=150, bbox_inches="tight")
plt.show()
```
